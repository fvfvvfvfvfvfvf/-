```python
import requests # подключает библиоткеку

# Замени на свой API-ключ
API_KEY = "eb4aef3a7bc85c912c1ac4c47d61ab7d"

def get_temperature(city_name):
    
    url = "https://api.openweathermap.org/data/2.5/weather"
    
    params = {
        "q": city_name,        # запрос 
        "appid": API_KEY,      # API-ключ
        "units": "metric",     # единицы измерения
        "lang": "ru"           # язык описания погоды
    }
    
    try:
        response = requests.get(url, params=params, timeout=10) #try защищает от сбоев, requests.get шлёт запрос с таймаутом, response.json() превращает ответ в словарь
        data = response.json()
        
        # Проверка на ошибки от API
        if response.status_code != 200:
            print(f"❌ Ошибка: {data.get('message', 'Неизвестная ошибка')}")
            return None
        
        # Извлекаем температуру
        temp = data["main"]["temp"] #встроенный словарь
        feels_like = data["main"]["feels_like"]
        description = data["weather"][0]["description"] #1.получаем список.2. выбираем первый елемент.3. берем текстовое описание
        
        print(f"\n🌍 Погода в городе: {city_name.capitalize()}")#все в фигцрных скобках заменяется на значение
        print(f"🌡️  Температура: {temp:.1f}°C")
        print(f"🤔 Ощущается как: {feels_like:.1f}°C")
        print(f"☁️  {description.capitalize()}")
        return temp
        
    except requests.exceptions.ConnectionError:
        print("❌ Ошибка: нет подключения к интернету") #except используется для того что бы выводить не красное поле ошибки а то что мы запрогромируем
    except requests.exceptions.Timeout:
        print("❌ Ошибка: превышено время ожидания")
    except Exception as e:
        print(f"❌ Непредвиденная ошибка: {e}")
    
    return None

if __name__ == "__main__":
    city = input("Введите название города (например, Moscow): ").strip() #что бы код выполняся при запуске а не импорте
    
    if city:
        get_temperature(city)
    else:
        print("❌ Название города не может быть пустым")
```
